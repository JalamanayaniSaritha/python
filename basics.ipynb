```python
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5

elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)

else:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)

print("Your bill amount: Rs", bill)
```

```python
# bank pin attempts
c_pin="7675"
attempt=0
max_attempts=3
while attempt<max_attempts:
    pin=input("enter your 4 digit pin")
    if pin==c_pin:
        print("acess granted")
        attempt=max_attempts
    else:
        print("incorrect pin")
        attempt+=1
if attempt==max_attempts and pin!=c_pin:
    print("account has been locked")
    
```
